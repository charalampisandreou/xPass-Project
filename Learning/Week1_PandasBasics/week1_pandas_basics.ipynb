```python
import pandas as pd
import requests

# URL for a single match event stream from StatsBomb open data 
# (Champions League Final: Barcelona vs Juventus)
match_url = "https://raw.githubusercontent.com/statsbomb/open-data/master/data/events/3754058.json"

# Download the raw nested JSON data stream
raw_events = requests.get(match_url).json()

# Convert the top-level array into a base dataframe
df_raw = pd.DataFrame(raw_events)

print("Raw Dataframe Dimensions:", df_raw.shape)
print("\nFirst 3 columns of the raw file:")
print(df_raw[['id', 'type', 'player']].head(5))
```

```python
df_raw['pass'].head
```

```python
df_raw.info()
```

```python
df_raw['type'].head
```

```python
df_raw['event_type'] = df_raw['type'].apply(lambda x: x.get('name') if isinstance(x, dict) else None)
```

```python
df_raw['event_type'].head
```

```python
df_pass = df_raw[df_raw['event_type'] == 'Pass']
df_pass.info()
```

```python
df_pass.iloc[0]
```

```python
df_pass['length'] = df_pass['pass'].apply(lambda x: x.get('length') if isinstance(x, dict) else None)
df_pass['length'].head()
```

```python
df_pass['angle'] = df_pass['pass'].apply(lambda x: x.get('angle') if isinstance(x, dict) else None)
df_pass['angle'].head()
```

```python
df_pass['height'] = df_pass['pass'].apply(lambda x: x.get('height',{}).get('name') if isinstance(x, dict) else None)
df_pass['height'].head()
```

```python
print(df_pass['pass'].iloc[101])
```

```python
df_pass['outcome'] = df_pass['pass'].apply(lambda x: x.get('outcome',{}).get('name') if isinstance(x, dict) else None)
df_pass['outcome'].head()
```

```python
df_pass['outcome'] = df_pass['outcome'].fillna('Complete')
df_pass['outcome'].head()
```

```python
df_pass['target'] = (df_pass['outcome'] == 'Complete').astype(int)
df_pass['target'].head()
```

```python
print(df_pass['target'].value_counts(normalize=True))
```

```python
df_pass['under_pressure'] = df_pass['under_pressure'].fillna(0).astype(int)
df_pass['under_pressure'].head()
```

```python
print(df_pass['under_pressure'].value_counts(normalize=True))
```

```python
print(df_pass.groupby('under_pressure')['target'].value_counts(normalize=True))
```
